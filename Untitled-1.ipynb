```python
1. What is Array?
Array has a fixed size, allows duplicate,Once we create an array, its size cannot be changed. 
Also, an array stores elements of a particular type.
int[] marks = new int[2];
marks[0] = 90;
marks[1] = 85;
But you cannot store a String:
marks[0] = "OOPS"; // Error
String[] names = new String[5];
What if tomorrow I have 10 students?

Collections Framework
                     |
       -----------------------------
       |            |              |
      List          Set           Queue
       |
 ArrayList
 LinkedList

Set
 |
HashSet
TreeSet

Map
 |
HashMap

Collections usually work with objects. Generics tell Java what type of object we want to store
2. ArrayList
ArrayList is dynamic in size. We can add or remove elements, so we don't have to decide the size in advance
particular datatype can be specified using Generics
ArrayList<Integer> marks = new ArrayList<>();
marks.add(10);
marks.add(20);
We didn't specify:
new ArrayList<>(5)
We simply created:
new ArrayList<>()
ArrayList<String> names = new ArrayList<>();
names.add("oops");
names.add("DBMS");
names.add("DS");
Size = number of elements currently stored.
Capacity = space available in the internal array

Generic class : at class level
class Student<T> {
    T data;

    Student(T data) {
        this.data = data;
    }

    void display() {
        System.out.println("Data: " + data);
    }
}

public class Main {
    public static void main(String[] args) {

        Student<Integer> s1 = new Student<>(100);
        s1.display();

        Student<String> s2 = new Student<>("OOPS");
        s2.display();

        Student<Double> s3 = new Student<>(85.5);
        s3.display();
    }
}

Generic Method: <T>: method level lo decide,pass different types to the same method
class Demo {
    public <T> void display(T data) {
        System.out.println(data);
    }
    public static void main(String[] args) {

        Demo d = new Demo();

        d.display(100);
        d.display("OOPS");
        d.display(85.5);
    }
}

Bounded type: set a rule about which data types are allowed
numeric types, such as Integer, Double, and Float 
int[] marks = {85, 95, 75};
Arrays.sort(marks);
Output:
[75, 85, 95]
-------------------------------------------------------------------------------------------------------
But what if we want to sort Student objects?
Each object contains a name and marks. Java needs to know which property to use for sorting.
Should it sort by name or by marks?
This is where Comparable and Comparator are useful.

** Comparable
import java.util.ArrayList;
import java.util.Collections;

class Student implements Comparable<Student> {

    String name;
    int marks;

    Student(String name, int marks) {
        this.name = name;
        this.marks = marks;
    }

    public int compareTo(Student s) {
        return this.marks - s.marks; or  return Integer.compare(this.marks, s.marks);
    }
}

public class Main {
    public static void main(String[] args) {

        ArrayList<Student> students = new ArrayList<>();

        students.add(new Student("Kavya", 85));
        students.add(new Student("Reddy", 95));
        students.add(new Student("Kinni", 75));

        Collections.sort(students);

        for (Student s : students) {
            System.out.println(s.name + " " + s.marks);
        }
    }
}
Return value	Meaning
Negative	   Current object comes before the other object
Zero	       Both objects are equal for this comparison
Positive	   Current object comes after the other object

**What is Comparator?
Suppose our Student class already sorts students by marks using Comparable.
Now we want to sort the same students by name instead.
We can use Comparator without changing the Student class.
import java.util.ArrayList;
import java.util.Collections;
import java.util.Comparator;

class Student {
    String name;
    int marks;

    Student(String name, int marks) {
        this.name = name;
        this.marks = marks;
    }
}

class NameComparator implements Comparator<Student> {

    public int compare(Student s1, Student s2) {
        return s1.name.compareTo(s2.name);//compares their names alphabetically
    }
}

public class Main {
    public static void main(String[] args) {

        ArrayList<Student> students = new ArrayList<>();

        students.add(new Student("Kavya", 85));
        students.add(new Student("Reddy", 95));
        students.add(new Student("kinni", 75));

        Collections.sort(students, new NameComparator());

        for (Student s : students) {
            System.out.println(s.name + " " + s.marks);
        }
    }
} 
```
